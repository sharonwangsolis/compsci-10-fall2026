### Control your computations with an `if` statement

```python
def sign(x): #user-defined function, not pre-built into the python language
    """determines the sign (+/-) of a numeric input"""
    # if-elif statements/branches below, because they are checking related categories
    # versus if we are checking if a number is negative, or it's even, or it's a three-digit number
    # that would be three different categories, we would use three separate if statements/branches
    if x > 0: # condition 1, this line is called an if statement
        return 'Positive' # lines 6-7 is called a branch, if a condition evalues to True, we "enter" a branch
        # meaning we see line 6 evaluates to true, we are granted permission to enter the branch, and evaluate line 7
    
    elif x < 0: # condition 2, #elif is short for else if
        return 'Negative'
    
    elif x == 0: # condition 3, == is checking for equality, is 0 == 0? True
        return 'Neither positive nor negative'
```

```python
sign(1)
```

```python
sign(-1)
```

```python
sign(0)
```

### Control your computations with a `for` loop - examples

Iterating over an array in-place

```python
#random selection
dice = make_array(1,2,3,4,5,6) # np.arange(1, 7) # remember we will start with 1 and go up to but not include 7
dice
```

```python
np.random.choice(dice) # simulate ONE dice roll
# each time i run this cell, i likely get a different answer
# this is also the first time we've seen something like in this class
```

```python
np.random.choice(dice, 10) #simulate TEN dice rolls
```

```python
np.random.choice(dice, 10) == 6 # check how many 6s I rolled in my ten rolls
```

```python
sum(np.random.choice(dice, 10) == 6) # check how many 6s I rolled in my ten rolls
# when we take a sum of an array, inside is bool values
# Python converts False = 0
# and True = 1
# we end up counting the number of 1s when we take the sum
# which represents the number of times we rolled the thing we indicated
# in this case, 6
# it answers our question, how many times did i roll a 6?
```
